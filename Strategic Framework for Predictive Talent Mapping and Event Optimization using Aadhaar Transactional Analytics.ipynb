## Problem Statement: 
National upskilling initiatives and hackathons often lack data-driven insights for optimal planning and location selection. Traditional approaches fail to accurately identify where youth populations are most active and which sectors align with regional industrial strengths. This results in inefficient resource allocation and missed opportunities to connect upskilling programs with local industry needs and startup ecosystems.

## Approach:
This project develops a data-driven framework that combines multiple datasets to optimize hackathon planning:

**Demand-Side Analysis:**

- Leverages high-frequency Aadhaar transactional data
- Identifies the "Active Youth Pipeline" by tracking spikes in:
    - Mandatory Biometric Updates (MBU) at age 15
    - Demographic Corrections at age 17
- Provides granular, district-level youth activity insights

**Supply-Side Integration:**

- AISHE College/University Dataset
- DPIIT Recognized Startup Data
- Major Industry Information (district level)

**Output:**

- Identifies sectoral trends at the district level
- Aligns State/District objectives with industry presence
- Enables targeted youth upskilling through industry-personal connections
- Provides smarter, data-backed solutions for hackathon planning and resource deployment

## Major Mathematical Framework & Administrative Assumptions:

### Core Assumptions:

* **The Age 15 MBU:** Mandatory Biometric Updates (MBU) are legally required at age 15. We assume spikes in the `Age_5_17` biometric dataset (minus new enrolments) represent the core 15-year-old "Youth Pipeline."
* **Board Exam Alignment:** Demographic updates (name/address/DoB) peaking at `Age_17` are assumed to be triggered by students aligning Aadhaar details with Class 10/12 certificates for college admissions.
* **The Formalization Gap:** New enrolments at `Age_18` are assumed to represent youth entering the formal economy (opening bank accounts or applying for scholarships) for the first time.
* **High Saturation Proxy:** With Jan 2026 Aadhaar saturation at **~95%**, we assume transactional density is a high-fidelity proxy for actual population distribution and activity.

---

### The Youth Population Index ($YPI$)
The $YPI$ quantifies the density of "Active" youth in a district. It measures real-world engagement and intent rather than just a static headcount.

$$YPI = \frac{(B_{5-17} - E_{5-17}) + (1.2 \times D_{17}) + E_{18}}{P_{saturated}} \times 100$$

**Breakdown:**

* **$(B_{5-17} - E_{5-17})$:** Isolates the mandatory update spike to identify the Age 15 pipeline.
* **$1.2 \times D_{17}$:** Demographic updates, weighted higher ($1.2$) due to their strong correlation with migration and higher education intent.
* **$E_{18}$:** New adult entrants, signaling digital and financial inclusion.
* **$P_{saturated}$:** Normalization factor based on Jan 2026 district saturation benchmarks.

---

### The Venue Success Score ($VSS$)
The $VSS$ is the final decision metric used to rank districts for a specific hackathon or upskilling theme.

$$VSS = (\omega_1 \times YPI) + (\omega_2 \times T_{density}) + (\omega_3 \times I_{base}) + (\omega_4 \times S_{hub})$$

**Explanation of Weightage:**

* **$\omega_1 \cdot YPI$ (30%):** Ensures a high volume of active, "identity-ready" youth.
* **$\omega_2 \cdot T_{density}$ (30%):** Matches district talent (AISHE college data) with the specific theme.
* **$\omega_3 \cdot I_{base}$ (20%):** Confirms availability of local industrial partners for mentorship and placement.
* **$\omega_4 \cdot S_{hub}$ (20%):** Validates the existing startup culture and innovation sentiment in that sector.

* **Adulthood Transition Pivot:** Demographic updates for the **Age $\geq$ 18** category show massive volume (exceeding 120k), illustrating a **"Transition Pivot"** where youth actively modify their profiles for college admissions, voter registration, or job placements.
* **Saturation Verification:** The near-zero enrolment rate for the **Age $\geq$ 18** category confirms that Mumbai has reached near-total Aadhaar saturation, meaning nearly all activity in this group represents **established youth** engaging with the digital ecosystem.
* **Concentrated Youth Activity:** The significant volume of Biometric Updates for **Age $\geq$ 18** ($\approx$ 280k) suggests that youth in major hubs perform comprehensive "document hygiene" to ensure their digital identity is ready for formal professional and educational opportunities.



---

### **The Youth Pulse Index ($YPI$) Formula**

Given that the statements above hold true, the following equation does not aim to provide an accurate total population headcount. Instead, it calculates a **"Pulse Count"** of individuals who are **actively engaging** with the Aadhaar ecosystem during critical life-stage transitions:

$$YPI = \frac{(B_{5-17} - E_{5-17}) + 1.2 \cdot D_{17} + E_{18}}{100}$$

> **Note:** This index prioritizes "Active" youth. While the census counts everyone, this equation identifies the **Market-Ready Youth** who are currently updating documents for education, employment, and formalization.

### Calculation of YPI(Youth Population Index):

```python
enrolment_grp = (
    enrolment_df
    .groupby(["State", "District"], as_index=False)
    .sum(numeric_only=True).drop(columns=['Pincode'])
)
enrolment_grp.columns = ["State", "District", "E_Age_0_5", "E_Age_5_17", "E_Age_18"]


demographic_grp = (
    demographic_df
    .groupby(["State", "District"], as_index=False)
    .sum(numeric_only=True).drop(columns=['Pincode'])
)
demographic_grp.columns = ["State", "District", "D_Age_5_17", "D_Age_18"]

biometric_grp = (
    biometric_df
    .groupby(["State", "District"], as_index=False)
    .sum(numeric_only=True).drop(columns=['Pincode'])
)
biometric_grp.columns = ["State", "District", "B_Age_5_17", "B_Age_18"]

YPI = (
    enrolment_grp
    .merge(demographic_grp, on=["State", "District"], how="inner")
    .merge(biometric_grp, on=["State", "District"], how="inner")
)
# Calculate Raw Youth Index
YPI['Youth_Index_Raw'] = ((YPI['B_Age_5_17'] - YPI['E_Age_5_17']) + 1.2 * YPI['D_Age_18'] + YPI['E_Age_18']) / 100

# Normalize to 0-1 Range (The "Youth Population Score") [ Min Max Normalization ]
YPI['Youth_Score'] = (YPI['Youth_Index_Raw'] - YPI['Youth_Index_Raw'].min()) / (YPI['Youth_Index_Raw'].max() - YPI['Youth_Index_Raw'].min())

# Reverse Saturation for Population Estimation
# (Using 95% as a general baseline for Jan 2026)
saturation_coeff = 0.95 
total_aadhaar_transactions = YPI['E_Age_18'] + YPI['D_Age_18'] + YPI['B_Age_18'] # Proxy for adult base

YPI['Est_Total_Pop'] = total_aadhaar_transactions / saturation_coeff
YPI['Projected_Youth_Headcount'] = YPI['Est_Total_Pop'] * 0.28

# Final Sorting
YPI = YPI.sort_values(by='Youth_Score', ascending=False)

### Uncomment this for raw print 
#print(YI[['State', 'District', 'Youth_Score', 'Projected_Youth_Headcount']].head(10))
YPI[['State', 'District', 'Youth_Score', 'Projected_Youth_Headcount']].head()
```

## Analyze and Assign the *Final_Score* for each sector using YPI, Industry, AISHE and Startup datasets:
Current PreAssigned Themes available: **FinTech, HealthTech, Agri_Innovation, Smart_Manufacturing, Fashion_Retail**

```python
college_df = aishe_df.groupby(['State', 'District', 'College_Category']).size().unstack(fill_value=0).reset_index()

# Special Case: Normalize District names to handle "Bangalore Urban" vs "Bengaluru" inconsistencies
def normalize_names(df, col):
    df[col] = df[col].str.lower().str.replace(' urban', '').str.replace(' rural', '').str.strip()
    return df

ypi_df = normalize_names(YPI, 'District')
college_df = normalize_names(college_df, 'District')
industry_df = normalize_names(industry_df, 'District')

# Consolidated Data Frame
master_df = ypi_df.merge(college_df, on=['State', 'District'], how='left') \
                 .merge(industry_df, on='District', how='left') \
                 .fillna(0)


def hackathon_planner(theme, target_state=None):
    """
    Inputs: 
    theme: ['FinTech', 'HealthTech', 'Smart_Manufacturing', 'Agri_Innovation']
    target_state: Filter for specific state (e.g., 'Tamil Nadu')
    """
    
    # Theme-Specific Weights and Mappings
    # Format: {Theme: (Related_Talent_Column, Related_Industry_Column, Related_Startup_Industry)}
    theme_map = {
        'FinTech': ('Business_Arts', 'Other', 'Finance Technology'),
        'HealthTech': ('Health_Science', 'Pharmaceticals', 'Healthcare & Lifesciences'),
        'Smart_Manufacturing': ('Engineering_Tech', 'Steel', 'Robotics'),
        'Agri_Innovation': ('Agri_Science', 'Rice Mills', 'Agriculture'),
        'Fashion_Retail': ('General', 'Garments', 'Fashion')
    }

    if theme not in theme_map:
        return "Theme not found. Please choose from: " + str(list(theme_map.keys()))

    tal_col, ind_col, star_ind = theme_map[theme]

    # Calculate Startup Density for that theme per State (from startup_df)
    state_startup_scores = startup_df[startup_df['Industry'] == star_ind].groupby('State')['Startups_Recognized'].sum().to_dict()

    # Step-by-Step Scoring
    df = master_df.copy()
    if target_state:
        df = df[df['State'] == target_state]

    # Normalize metrics (Min-Max Scaling) to 0-1
    def scale(series):
        return (series - series.min()) / (series.max() - series.min() + 1e-9)

    # SUCCESS SCORE FORMULA
    # 30% Youth Activity (Aadhaar) + 30% Local Talent (AISHE) + 20% Industry base + 20% Startup Hub
    df['Startup_Score'] = df['State'].map(state_startup_scores).fillna(0)
    
    df['Final_Score'] = (
        0.30 * scale(df['Youth_Score']) + 
        0.30 * scale(df[tal_col]) + 
        0.20 * scale(df[ind_col]) + 
        0.20 * scale(df['Startup_Score'])
    )

    results = df.sort_values(by='Final_Score', ascending=False).head(5)
    
    print(f"\n--- TOP 5 LOCATIONS FOR {theme.upper()} HACKATHON ---")
    return results[['State', 'District', 'Youth_Score', tal_col, ind_col, 'Final_Score']]

```

# Result:
- What are the **Best Places** to conduct a **Smart Manufacturing Hackathon** in **Tamil Nadu**? Provided the event held should help participants collaborate with industry, startup professionals and in overall help the students preparation and increase participation.

```python
### Usage Example
# To find the best place for a Smart Manufacturing event in Tamil Nadu:
results = hackathon_planner('Smart_Manufacturing', target_state='Tamil Nadu')
gdf = gpd.read_file("./india-maps-data/topojson/states/tamilnadu.json", layer=district_layer)
gdf['district'] = gdf['district'].str.lower()
merged_map = gdf.merge(results, left_on="district", right_on="District", how="left")


fig, ax = plt.subplots(figsize=(7,7))
merged_map.plot(column="Final_Score", cmap="OrRd", legend=True, missing_kwds={"color": "gray"}, ax=ax)
plt.show()
results
```

# Anomaly Detection and Pattern Recognition
While the primary focus of this framework is predictive talent mapping, the Aadhaar dataset reveals several interesting anomalies that warrant investigation and provide additional insights into youth engagement patterns across India.

## Key Anomalies Detected

1. **High Demographic, Low Biometric Updates**

**Pattern:** Districts with disproportionately high demographic updates (Age 17) but low biometric updates (Age 5-17)

**Interpretation:**
  
- Indicates migration hotspots where youth are updating addresses for college admissions
- Suggests incoming student population from other districts
- Example districts showing this pattern: Pune, Bengaluru Urban, Hyderabad

```python
# Identify districts where Demographic Updates > 2x Biometric Updates
migration_hotspots = master_df[
    (master_df['D_Age_18'] > 2 * master_df['B_Age_5_17'])
].sort_values(by='D_Age_18', ascending=False)

print("Detected Migration Hotspots (Potential Educational Hubs):")
print(migration_hotspots[['State', 'District', 'D_Age_18', 'B_Age_5_17']].head())
```

2. **Low Youth Index Despite High College Density**

**Pattern:** Districts with many educational institutions but surprisingly low YPI scores

**Interpretation:**

- May indicate commuter student populations who maintain Aadhaar records in home districts
- Suggests college infrastructure exists but local youth engagement is lower than expected
- Potential digital literacy gaps or Aadhaar center accessibility issues

```python
# High College Count but Bottom 25% YPI
def normalize_geo(df):
    for col in ['State', 'District']:
        if col in df.columns:
            df[col] = df[col].astype(str).str.lower().str.strip()
            df[col] = df[col].str.replace(' urban', '').str.replace(' rural', '')
    return df

ypi_df = normalize_geo(ypi_df)
college_df = normalize_geo(college_df)
industry_df = normalize_geo(industry_df)
aishe_raw = normalize_geo(aishe_df)

college_counts = aishe_raw.groupby('District').size().reset_index(name='College_Count')

master_df = ypi_df.merge(college_df, on=['State', 'District'], how='left') \
                 .merge(industry_df, on='District', how='left') \
                 .merge(college_counts, on='District', how='left') \
                 .fillna(0)

talent_gap_districts = master_df[
    (master_df['College_Count'] > master_df['College_Count'].median()) & 
    (master_df['Youth_Score'] < master_df['Youth_Score'].quantile(0.25))
]

print("\nInfrastructure-Engagement Mismatch (Potential Digital Literacy Gaps):")
print(talent_gap_districts[['State', 'District', 'College_Count', 'Youth_Score']].head())
```

3. **Sudden Enrolment Spikes in Age 18 Category**

**Pattern:** Districts showing abnormally high new enrolments for Age ≥18 compared to national average

**Interpretation:**

- Represents formalization frontier - youth entering banking/employment systems for first time
- Often seen in tier-2/tier-3 cities experiencing rapid economic development
- Indicates emerging talent pools not yet captured by traditional university metrics

```python
# Detect districts with Age 18 Enrolment > 3 Standard Deviations from Mean
mean_e18 = master_df['E_Age_18'].mean()
std_e18 = master_df['E_Age_18'].std()

formalization_frontier = master_df[
    master_df['E_Age_18'] > (mean_e18 + 3 * std_e18)
]

print("Formalization Frontier (Rapidly Developing Emerging Pools):")
print(formalization_frontier[['State', 'District', 'E_Age_18']].head())
```

# Limitations and Future Work

## Current Limitations

### Data Constraints

**Transactional vs. Saturated Population Data**

- The Aadhaar dataset provides transaction counts (enrolments and updates) rather than total saturated population per district
- Limits our ability to calculate true per-capita participation rates, only relative activity indices
- Transaction density serves as a proxy for youth activity, which may be influenced by regional digital literacy variations and administrative drives

**Limited Temporal Coverage**

- Analysis constrained to a single year (2025-2026) of Aadhaar data
- Cannot identify multi-year trends, seasonal patterns in academic cycles, or long-term migration trajectories
- Temporal snapshot may not represent consistent annual patterns across all districts

**District Name Standardization**

- Inconsistent naming conventions across datasets required manual normalization (e.g., "Bangalore Urban" vs. "Bengaluru")
- Potential for mismatches in district-level aggregations and loss of granular data for newly formed administrative units

### Methodological Limitations

**Fixed Weight Assignment in VSS Formula**

- Venue Success Score weights (30% Youth, 30% Talent, 20% Industry, 20% Startup) are based on logical reasoning rather than empirical validation
- Same weights applied uniformly across all themes, though optimal distributions likely vary by sector
- No ground truth data available to test predictive accuracy

**Saturation Coefficient Assumption**

- 95% Aadhaar saturation rate applied uniformly across all districts
- Ignores documented regional disparities (urban vs. rural, geographic variations)
- Lacks district-specific saturation benchmarks for precise population estimation

**College Categorization Simplification**

- Keyword-based matching may misclassify institutions with non-standard naming
- "General" category serves as catch-all, reducing analytical precision
- Cannot detect interdisciplinary programs spanning multiple categories

### Validation Gaps

- Framework not validated against actual participation rates from past government hackathons
- No testing against post-event satisfaction surveys or industry engagement metrics
- Unable to measure prediction error rates or establish confidence intervals without historical outcome data
